```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
```

```python
plt.imshow(data.coffee())
```

```python
def implot(img, img_filtrada, kernelsize):
  plt.subplot(1, 2, 1)
  plt.imshow(img)
  plt.title('Original')
  plt.axis('off')

  plt.subplot(1, 2, 2)
  plt.imshow(img_filtrada)
  plt.title('Filtrada' + kernelsize)
  plt.axis('off')

```

```python
img = data.coffee()

img_filtrada = cv2.blur(img, (5, 5))
plt.figure(figsize=(10, 10))

implot(img, img_filtrada, '5x5')
```

```python
kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
])


img_filtrada = cv2.filter2D(img, -1, kernel)
implot(img, img_filtrada, '3x3')
```
