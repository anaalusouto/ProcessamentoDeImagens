```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = 'papoi.jpg'

original_color_image = cv2.imread(image_path)
if original_color_image is None:
    raise FileNotFoundError(f"Não foi possível carregar a imagem em: {image_path}")

grayscale_image = cv2.cvtColor(original_color_image, cv2.COLOR_BGR2GRAY)

print(f"Imagem '{image_path}' carregada e convertida para tons de cinza. Dimensões: {grayscale_image.shape}")
```

```python
def quantize_image(image, bits):
    levels = 2**bits

    if bits == 8:
        return image

    scaled_image = (image / 255.0) * (levels - 1)

    quantized_scaled_image = np.round(scaled_image)

    quantized_image = (quantized_scaled_image / (levels - 1)) * 255

    return quantized_image.astype(np.uint8)
```

```python
quantized_1_bit = quantize_image(grayscale_image, 1)
quantized_2_bits = quantize_image(grayscale_image, 2)
quantized_4_bits = quantize_image(grayscale_image, 4)
quantized_8_bits = quantize_image(grayscale_image, 8)

print("Imagens quantizadas geradas para 1, 2, 4 e 8 bits.")
```

```python
fig, axes = plt.subplots(1, 5, figsize=(20, 4))

axes[0].imshow(grayscale_image, cmap='gray')
axes[0].set_title('Original (Tons de Cinza)')
axes[0].axis('off')

axes[1].imshow(quantized_1_bit, cmap='gray')
axes[1].set_title('1 bit (2 Níveis)')
axes[1].axis('off')

axes[2].imshow(quantized_2_bits, cmap='gray')
axes[2].set_title('2 bits (4 Níveis)')
axes[2].axis('off')

axes[3].imshow(quantized_4_bits, cmap='gray')
axes[3].set_title('4 bits (16 Níveis)')
axes[3].axis('off')

axes[4].imshow(quantized_8_bits, cmap='gray')
axes[4].set_title('8 bits (256 Níveis)')
axes[4].axis('off')

plt.suptitle('Quantização de Imagens em Tons de Cinza', fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])

plt.savefig('quantizacao.png')
plt.show()

print("Figura 'quantizacao.png' salva com sucesso.")
```
