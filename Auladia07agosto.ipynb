```python
img = cv2.imread('4f67eb36e32736b9adaffe127a592e42.jpg')
B, G, R = cv2.split(img)
cv2_imshow(B)
```

```python
plt.subplot(1,3,1)
plt.imshow(B)
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(G)
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(R)
plt.axis('off')
```

```python
plt.subplot(1,3,1)
plt.imshow(B, cmap='grey')
plt.axis('off')
plt.title('Canal azul')

plt.subplot(1,3,2)
plt.imshow(G, cmap='grey')
plt.axis('off')
plt.title('Canal verde')

plt.subplot(1,3,3)
plt.imshow(R, cmap='grey')
plt.axis('off')
plt.title('Canal vermelho')

```
