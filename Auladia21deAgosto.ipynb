```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
img = cv2.imread('prismo.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

img_rgb
```

```python
plt.figure(figsize=(5,3))
plt.imshow(img)
plt.title('Original Image')
plt.show()
```

```python
fig, ax = plt.subplots(2,2, figsize=(6,4))
ax[0,0].imshow(img)
ax[0,0].set_title('BGR image')
ax[0,0].axis('off')

ax[0,1].imshow(img_rgb)
ax[0,1].set_title('RGB Image')
ax[0,1].axis('off')

ax[1,0].imshow(img_hsv)
ax[1,0].set_title('HSV Image')
ax[1,0].axis('off')

ax[1,1].imshow(img_gray, cmap='gray')
ax[1,1].set_title('Gray Image')
ax[1,1].axis('off')

plt.tight_layout()

plt.savefig('image_types.png', dpi=300, bbox_inches='tight')
plt.show()
```
